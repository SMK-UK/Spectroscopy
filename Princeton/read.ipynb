```python
from SPE2map import SPE2map
import numpy as np
import matplotlib.pyplot as mp
from pathlib import Path
import os

calibration = 605.476
directory = r"C:\Users\Sean\Desktop\L405P20\Spectra"
path_cal = r"C:\Users\Sean\Desktop\L405P20\Spectra\Calibration\606_cal.SPE"
             
cal = SPE2map(path_cal)
waves, intensity = cal.getSpectra()
peak = np.argmax(intensity) 
cal_value = calibration - waves[peak]

fig, ax = mp.subplots()
ax.plot(waves, intensity)
ax.set(title='Uncalibrated',
       xlabel='Wavelength (nm)',
       ylabel='Counts (Hz)',
       xlim=[603, 606])
```

```python
files = [f for f in Path(directory).iterdir() if f.is_file()]
data = [SPE2map(x).getSpectra() for x in files]

for names, values in zip(files, data):

    peak_idx = np.argmax(values[1])
    name = names.stem
    waves = values[0]+cal_value
    peak = np.round(waves[peak_idx],2)

    fig, ax = mp.subplots()
    ax.plot(waves, values[1])
    ax.set(
        title=f"{name}, peak @ {peak}nm",
        xlabel='Wavelength (nm)',
        ylabel='Counts (Hz)'#,
        #xlim=[peak-2.5, peak+2.5]
    )
    print(name, peak)
```
